# BRONZE LAYER

### Step:- 1 Define a Schema 

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DoubleType


sales_schema = StructType([
    StructField("Sales_ID", StringType(), True),
    StructField("SKU_ID", StringType(), True),
    StructField("Sale_Date", StringType(), True),
    StructField("Store_ID", StringType(), True),
    StructField("Sales_Channel", StringType(), True),
    StructField("Customer_ID", StringType(), True),
    StructField("Quantity", IntegerType(), True),
    StructField("Unit_Price", DoubleType(), True),
    StructField("Discount", DoubleType(), True),
    StructField("Final_Amount", DoubleType(), True),
    StructField("Payment_Method", StringType(), True)
])

secondary_market_schema = StructType([
    StructField("SecondaryMarket_ID", StringType(), True),
    StructField("SKU_ID", StringType(), True),
    StructField("Original_Purchase_Date", StringType(), True),
    StructField("Resale_Date", StringType(), True),
    StructField("Resale_Price", DoubleType(), True),
    StructField("Condition", StringType(), True),
    StructField("Seller_Type", StringType(), True),
    StructField("Buyer_Type", StringType(), True),
    StructField("Platform", StringType(), True),
    StructField("City", StringType(), True),
    StructField("Transaction_Status", StringType(), True)
])

manufacturing_schema = StructType([
    StructField("SKU_ID", StringType(), True),
    StructField("Product_Name", StringType(), True),
    StructField("Brand", StringType(), True),
    StructField("Model", StringType(), True),
    StructField("Manufacturing_Date", StringType(), True),
    StructField("Component", StringType(), True),
    StructField("Component_Cost", DoubleType(), True),
    StructField("Warranty_Period", IntegerType(), True),
    StructField("Failure_Rate", DoubleType(), True),
    StructField("Manufacturing_Cost", DoubleType(), True)
])

warranty_schema = StructType([
    StructField("SKU_ID", StringType(), True),
    StructField("Product_Name", StringType(), True),
    StructField("Brand", StringType(), True),
    StructField("Model", StringType(), True),
    StructField("Manufacturing_Date", StringType(), True),
    StructField("Component", StringType(), True),
    StructField("Component_Cost", DoubleType(), True),
    StructField("Warranty_Period", IntegerType(), True),
    StructField("Failure_Rate", DoubleType(), True),
    StructField("Manufacturing_Cost", DoubleType(), True)
])
```

### Step 2 Read as CSV


```python
sales_bronze_df = spark.read.format("csv") \
    .option("header", "true") \
    .schema(sales_schema) \
    .load("/Volumes/echo_chain_proj-1/raw_data/raw_file/Laptop_Sales_Dataset_10000.csv")



manufacturing_bronze_df = spark.read.format("csv") \
    .option("header", "true") \
    .schema(manufacturing_schema) \
    .load("/Volumes/echo_chain_proj-1/raw_data/raw_file/Manufacturing_Dataset_10000.csv")



warranty_bronze_df = spark.read.format("csv") \
    .option("header", "true") \
    .schema(warranty_schema) \
    .load("/Volumes/echo_chain_proj-1/raw_data/raw_file/Warranty_Dataset_5000.csv")



secondary_market_bronze_df = spark.read.format("csv") \
    .option("header", "true") \
    .schema(secondary_market_schema) \
    .load("/Volumes/echo_chain_proj-1/raw_data/raw_file/Laptop_Secondary_Market_Dataset_10000.csv")
```

```python
sales_bronze_df .display()
secondary_market_bronze_df.display()
manufacturing_bronze_df.display()
warranty_bronze_df.display()
```

### Step - 3 Add Ingestion metadata

```python
from pyspark.sql.functions import current_timestamp

sales_bronze_df= sales_bronze_df.withColumn('ingest_timestamp', current_timestamp())

secondary_market_bronze_df = secondary_market_bronze_df.withColumn('ingest_timestamp',current_timestamp())

manufacturing_bronze_df=manufacturing_bronze_df.withColumn('ingest_timestamp', current_timestamp())

warranty_bronze_df = warranty_bronze_df.withColumn('ingest_timestamp',current_timestamp())
```

### Step -4 Save AS Delta Table in Bronze

```python
sales_bronze_df.write.format("delta").mode("overwrite").saveAsTable("`echo_chain_proj-1`.bronze.bronze_sales")

secondary_market_bronze_df .write.format("delta").mode("overwrite").saveAsTable("`echo_chain_proj-1`.bronze.bronze_secondary_market")

manufacturing_bronze_df.write.format("delta").mode("overwrite").saveAsTable("`echo_chain_proj-1`.bronze.bronze_manufacturing")

warranty_bronze_df  .write.format("delta").mode("overwrite").saveAsTable("`echo_chain_proj-1`.bronze.bronze_warranty")

```
