# Bivarites Analysis


```python
sales_df.groupby("Sales_Channel")["Final_Amount"].sum()
```

```python
sns.barplot(
    x="Sales_Channel",
    y="Final_Amount",
    data=sales_df,
    estimator=sum
)

plt.xticks(rotation=45)
plt.show()
```
