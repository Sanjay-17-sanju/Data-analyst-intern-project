### String to float to int (sales)

```python
sales_df["Final_Amount"] = (
    sales_df["Final_Amount"]
    .str.replace(",", "", regex=False)
)

sales_df["Final_Amount"] = pd.to_numeric(
    sales_df["Final_Amount"],
    errors="coerce"
)
```

```python
sales_df["Final_Amount"] = sales_df["Final_Amount"].astype(int)
```

```python
sales_df["Final_Amount"].dtype
```

```python
secondary_df["Resale_Price"] = (
    secondary_df["Resale_Price"]
    .str.replace(",", "", regex=False)
)

secondary_df["Resale_Price"] = pd.to_numeric(
    secondary_df["Resale_Price"],
    errors="coerce"
)
```

```python
secondary_df["Resale_Price"] = secondary_df["Resale_Price"].astype(int)
```
