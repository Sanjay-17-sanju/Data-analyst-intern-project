### string to float to int (maunufacturing)

```python
manufacturing_df["Manufacturing Cost"].dtype
```

```python
manufacturing_df["Manufacturing Cost"] = (
    manufacturing_df["Manufacturing Cost"]
    .str.replace(",", "", regex=False)
)

manufacturing_df["Manufacturing Cost"] = pd.to_numeric(
    manufacturing_df["Manufacturing Cost"],
    errors="coerce"
)
```

```python
manufacturing_df["Manufacturing Cost"] = manufacturing_df["Manufacturing Cost"].astype(int)
```

```python
manufacturing_df["Component Cost"].dtype
```

```python
manufacturing_df["Component Cost"] = (
    manufacturing_df["Component Cost"]
    .str.replace(",", "", regex=False)
)

manufacturing_df["Component Cost"] = pd.to_numeric(
    manufacturing_df["Component Cost"],
    errors="coerce"
)
```

```python
manufacturing_df["Component Cost"] = manufacturing_df["Component Cost"].astype(int)
```
