# Bivarites Analysis

## For Sales

### Sales_Channel + Payment_Method + Final_Amount = Revenue Analysis

```python
sales_df.groupby(
    ["Sales_Channel", "Payment_Method"]
)["Final_Amount"].sum().reset_index()
```

### For Secondary

### Condition + Platform + Resale_Price = Resale Analysis

```python
secondary_df.groupby(
    ["Condition", "Platform"]
)["Resale_Price"].sum().reset_index()
```

### For manufacturing

### Brand + Component + Failure_Rate = Failure Analysis

```python
manufacturing_df.groupby(
    ["Brand", "Component"]
)["Failure Rate"].mean().reset_index()
```

### For Warranty

```python
warranty_df.groupby(
    ["Brand", "Component"]
)["Failure Rate"].mean().reset_index()
```

### string to float to int(secondary)

```python
secondary_df["Resale_Price"].dtype
```

```python
secondary_df["Resale_Price"] = (
    secondary_df["Resale_Price"]
    .str.replace(",", "", regex=False)
)

secondary_df["Resale_Price"] = pd.to_numeric(
    secondary_df["Resale_Price"],
    errors="coerce"
)
```

```python
secondary_df["Resale_Price"] = secondary_df["Resale_Price"].astype(int)
```

### string to float to int (sales) 

```python
sales_df["Final_Amount"] = (
    sales_df["Final_Amount"]
    .str.replace(",", "", regex=False)
)

sales_df["Final_Amount"] = pd.to_numeric(
    sales_df["Final_Amount"],
    errors="coerce"
)
```

```python
sales_df["Final_Amount"] = sales_df["Final_Amount"].astype(int)
```
