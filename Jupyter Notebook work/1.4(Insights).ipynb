# Insights

# For Sales

### 1.Total Revenue

```python
print("Total Revenue:", sales_df["Final_Amount"].sum())
```

### 2.Average Revenue

```python
print("Average Revenue:", sales_df["Final_Amount"].mean())
```

### 3.Highest Sales Channel

```python
 sales_df.groupby("Sales_Channel")["Final_Amount"].sum().sort_values(ascending=False)
```

### 4.Best Payment Method

```python
payment_revenue = (
    sales_df.groupby("Payment_Method")["Final_Amount"]
    .sum()
    .sort_values(ascending=False)
)

print(payment_revenue)

```

### 6.Highest Selling SKU_ID

```python
sku_quantity = (
    sales_df.groupby("SKU_ID")["Quantity"]
    .sum()
    .sort_values(ascending=False)
)

print(sku_quantity)
```

```python
print("Highest Selling SKU:")
print(sku_quantity.head(1))
```

### String to float to int (manufacturing)

```python
manufacturing_df["Manufacturing Cost"].dtype
```

```python
manufacturing_df["Manufacturing Cost"] = (
    manufacturing_df["Manufacturing Cost"]
    .str.replace(",", "", regex=False)
)

manufacturing_df["Manufacturing Cost"] = pd.to_numeric(
    manufacturing_df["Manufacturing Cost"],
    errors="coerce"
)
```

```python
manufacturing_df["Manufacturing Cost"] = manufacturing_df["Manufacturing Cost"].astype(int)
```

### String to float to int (sales)

```python
sales_df["Final_Amount"] = (
    sales_df["Final_Amount"]
    .str.replace(",", "", regex=False)
)

sales_df["Final_Amount"] = pd.to_numeric(
    sales_df["Final_Amount"],
    errors="coerce"
)
```

```python
sales_df["Final_Amount"] = sales_df["Final_Amount"].astype(int)
```

```python
sales_df["Final_Amount"].dtype
```

### String to float to int (secondary)

```python
secondary_df["Resale_Price"] = (
    secondary_df["Resale_Price"]
    .str.replace(",", "", regex=False)
)

secondary_df["Resale_Price"] = pd.to_numeric(
    secondary_df["Resale_Price"],
    errors="coerce"
)
```

```python
secondary_df["Resale_Price"] = secondary_df["Resale_Price"].astype(int)
```
